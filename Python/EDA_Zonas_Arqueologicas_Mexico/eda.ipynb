Si el toral de visitantes sube entre 2000 y 2025, una parte puede deberse a que hay más sitios en el conteo y no solo a que hay más gente.

- No hay que eliminar nada, ya que no existen duplicados.

- Los 32 faltantes de clave_siinah no son aleatorios. Pertenecen a solo 6 recintos (Mixcoac, Rincón Colorado, Arroyo Seco, Ichkabal, Cerro del Teúl y Cerro de las Ventanas) que se repiten año tras año. Simplemente son sitios sin codigo asignado.

```python
# Creamos nuevas columnas en el dataframe que representen el total de visitantes por mes, 
# sumando los valores de las columnas correspondientes a visitantes nacionales y extranjeros

meses = ['enero', 'febrero', 'marzo', 'abril', 'mayo', 'junio', 'julio', 'agosto', 'septiembre', 'octubre', 'noviembre', 'diciembre']
for mes in meses:
    df[f'{mes}_total'] = df[f'{mes}_nac'] + df[f'{mes}_ext']
```

```python
# Creamos nuevas columnas en el dataframe que representen el total de visitantes nacionales y extranjeros, 
# sumando los valores de las columnas correspondientes a cada mes

cols_nac = [f'{mes}_nac' for mes in meses]
cols_ext = [f'{mes}_ext' for mes in meses]

df['total_nac'] = df[cols_nac].sum(axis=1)
df['total_ext'] = df[cols_ext].sum(axis=1)
df['total'] = df['total_nac'] + df['total_ext']

print('Total de visitantes nacionales:', df['total_nac'].sum())
print('Total de visitantes extranjeros:', df['total_ext'].sum())
```

```python
# Creamos un nuevo dataframe llamado df_completo que contenga solo las filas del dataframe original donde el año sea menor a 2026

df_completo = df[df['anio'] < 2026].copy()
print(df_completo.shape)
```

```python
# Mostramos un resumen estadístico de las columnas 'total_nac', 'total_ext' y 'total' del dataframe df_completo

df_completo[['total_nac', 'total_ext', 'total']].describe()
```

```python
# Calculamos el total de visitantes nacionales, extranjeros y el total general por año, agrupando los datos del dataframe df_completo por la columna 'anio'

totales_por_anio = df_completo.groupby('anio')[['total_nac', 'total_ext', 'total']]
print(totales_por_anio.sum())
```

```python
# Calculamos el número de años distintos en los que se registraron visitantes para cada recinto arqueológico, 
# agrupando los datos del dataframe df por la columna 'recinto' y contando los valores únicos de la columna 'anio'

anios_por_recinto = df.groupby('recinto')['anio'].nunique().sort_values()
print(anios_por_recinto.head(15))
```

El resultado de este último bloque resulta interesante, se presentan sitios nuevos y alta rotacion, pero Teotihuacan aparece solo un año, lo cual es sospechoso porque este es uno de los recintos mas visitados de México y es improbable que solo tenga datos de un año. Así que debemos verificar si existen otros sitios con la palabra Teotihuacan en su nombre.

Los recintos "Las Ventanas" y "Cerro de las Ventanas" tienen nombres parecidos, y puede que sean dos sitios distintos o uno solo con dos nombres.

```python
# ¿Qué nombres tiene Teotihuacan y en qué años aparece cada uno?

df[df['recinto'].str.contains('Teotihuacan', case=False)] \
    .groupby('recinto')['anio'].agg(['min', 'max', 'count'])
```

```python
# ¿Son lo mismo los dos "Ventanas"?

df[df['recinto'].str.contains('Ventanas', case=False)] \
    .groupby(['recinto', 'estado'])['anio'].agg(['min', 'max', 'count'])
```

```python
# Rango de años de los sitios con menos apariciones

df.groupby('recinto')['anio'].agg(['min', 'max', 'nunique']).sort_values('nunique').head(15)
```

Con este ultimo bloque nos podemos dar cuenta de que Teotihuacan realmente es un problema de captura ya que la unica diferencia son las mayusculas y minisculas y para pandas son textos distintos.
Asi que lo que debemos hacer es corregir este error y quedarnos solo con uno.

```python
# Corregimos el nombre del recinto "Zona Arqueologica Teotihuacan (con museo de sitio)" a "Zona Arqueologica Teotihuacan (con museo de Sitio)" 
# en la columna 'recinto' del dataframe df

df['recinto'] = df['recinto'].replace({
    'Zona Arqueologica Teotihuacan (con museo de sitio)':
    'Zona Arqueologica Teotihuacan (con museo de Sitio)'
})

print(df['recinto'].unique())
```

```python
# Mostramos los nombres únicos de los recintos arqueológicos en el dataframe df, convirtiéndolos a minúsculas y eliminando espacios en blanco al inicio y al final

print(df['recinto'].str.lower().str.strip().unique())
```

```python
# Mostramos las filas del dataframe df donde el nombre del recinto contenga la palabra "Ventanas", 
# incluyendo las columnas 'anio', 'recinto', 'clave_siinah' y 'total', y ordenadas por año

df[df['recinto'].str.contains('Ventanas')][['anio', 'recinto', 'clave_siinah', 'total']].sort_values('anio')
```

En este ultimo bloque analizamos que 'Las Ventanas' y 'Cerro de las Ventanas' no sean el mismo. Para comparamos el anio, recinto y clave_siinah.

Con todo esto tenemos dos criterios que nos indican que no son el mismo sitio, ya que las claves de ambos son distintas, ademas de que una tiene siempre cero visitantes y la otra tiene miles.

Pero algo que no podemos pasar desapercibido son los ceros ya que pueden signofocar que se registro cero visitantes o que no hay dato.
Podria ser un sitio que no estaba abierto al publico, o un sitio que no reporto y quedo en cero por defecto. 
Ejemplo: 'Cerro de las Ventanas' en 2021 tiene total de ceri visitantes, jsuto en pandemia, lo que podria encajar con un cierre temporal.
