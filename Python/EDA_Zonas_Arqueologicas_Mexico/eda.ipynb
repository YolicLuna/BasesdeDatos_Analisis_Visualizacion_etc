```python
# Creamos nuevas columnas en el dataframe que representen el total de visitantes por mes, 
# sumando los valores de las columnas correspondientes a visitantes nacionales y extranjeros

meses = ['enero', 'febrero', 'marzo', 'abril', 'mayo', 'junio', 'julio', 'agosto', 'septiembre', 'octubre', 'noviembre', 'diciembre']
for mes in meses:
    df[f'{mes}_total'] = df[f'{mes}_nac'] + df[f'{mes}_ext']
```

```python
# Creamos nuevas columnas en el dataframe que representen el total de visitantes nacionales y extranjeros, 
# sumando los valores de las columnas correspondientes a cada mes

cols_nac = [f'{mes}_nac' for mes in meses]
cols_ext = [f'{mes}_ext' for mes in meses]

df['total_nac'] = df[cols_nac].sum(axis=1)
df['total_ext'] = df[cols_ext].sum(axis=1)
df['total'] = df['total_nac'] + df['total_ext']

print('Total de visitantes nacionales:', df['total_nac'].sum())
print('Total de visitantes extranjeros:', df['total_ext'].sum())
```

```python
# Creamos un nuevo dataframe llamado df_completo que contenga solo las filas del dataframe original donde el año sea menor a 2026

df_completo = df[df['anio'] < 2026].copy()
print(df_completo.shape)
```

```python
# Mostramos un resumen estadístico de las columnas 'total_nac', 'total_ext' y 'total' del dataframe df_completo

df_completo[['total_nac', 'total_ext', 'total']].describe()
```

```python
# Calculamos el total de visitantes nacionales, extranjeros y el total general por año, agrupando los datos del dataframe df_completo por la columna 'anio'

totales_por_anio = df_completo.groupby('anio')[['total_nac', 'total_ext', 'total']].sum()
print(totales_por_anio)
```

```python
# Calculamos el número de años distintos en los que se registraron visitantes para cada recinto arqueológico, 
# agrupando los datos del dataframe df por la columna 'recinto' y contando los valores únicos de la columna 'anio'

anios_por_recinto = df.groupby('recinto')['anio'].nunique().sort_values()
print(anios_por_recinto.head(15))
```

El resultado de este último bloque resulta interesante, se presentan sitios nuevos y alta rotacion, pero Teotihuacan aparece solo un año, lo cual es sospechoso porque este es uno de los recintos mas visitados de México y es improbable que solo tenga datos de un año. Así que debemos verificar si existen otros sitios con la palabra Teotihuacan en su nombre.

Los recintos "Las Ventanas" y "Cerro de las Ventanas" tienen nombres parecidos, y puede que sean dos sitios distintos o uno solo con dos nombres.

```python
# ¿Qué nombres tiene Teotihuacan y en qué años aparece cada uno?

df[df['recinto'].str.contains('Teotihuacan', case=False)] \
    .groupby('recinto')['anio'].agg(['min', 'max', 'count'])
```

```python
# ¿Son lo mismo los dos "Ventanas"?

df[df['recinto'].str.contains('Ventanas', case=False)] \
    .groupby(['recinto', 'estado'])['anio'].agg(['min', 'max', 'count'])
```

```python
# Rango de años de los sitios con menos apariciones

df.groupby('recinto')['anio'].agg(['min', 'max', 'nunique']).sort_values('nunique').head(15)
```

Con este ultimo bloque nos podemos dar cuenta de que Teotihuacan realmente es un problema de captura ya que la unica diferencia son las mayusculas y minisculas y para pandas son textos distintos.
Asi que lo que debemos hacer es corregir este error y quedarnos solo con uno.

```python
# Corregimos el nombre del recinto "Zona Arqueologica Teotihuacan (con museo de sitio)" a "Zona Arqueologica Teotihuacan (con museo de Sitio)" 
# en la columna 'recinto' del dataframe df

df['recinto'] = df['recinto'].replace({
    'Zona Arqueológica Teotihuacan (con museo de sitio)':
    'Zona Arqueológica Teotihuacan (con Museo de Sitio)'
})

print(df['recinto'].unique())
```

```python
# Mostramos los nombres únicos de los recintos arqueológicos en el dataframe df, convirtiéndolos a minúsculas y eliminando espacios en blanco al inicio y al final

print(df['recinto'].str.lower().str.strip().unique())
```

```python
# Mostramos las filas del dataframe df donde el nombre del recinto contenga la palabra "Ventanas", 
# incluyendo las columnas 'anio', 'recinto', 'clave_siinah' y 'total', y ordenadas por año

df[df['recinto'].str.contains('Ventanas')][['anio', 'recinto', 'clave_siinah', 'total']].sort_values('anio')
```

En este ultimo bloque analizamos que 'Las Ventanas' y 'Cerro de las Ventanas' no sean el mismo. Para comparamos el año, recinto y clave_siinah.

Con todo esto tenemos dos criterios que nos indican que no son el mismo sitio, ya que las claves de ambos son distintas, ademas de que una tiene siempre cero visitantes y la otra tiene miles.

Pero algo que no podemos pasar desapercibido son los ceros ya que pueden signofocar que se registro cero visitantes o que no hay dato.
Podria ser un sitio que no estaba abierto al publico, o un sitio que no reporto y quedo en cero por defecto. 
Ejemplo: 'Cerro de las Ventanas' en 2021 tiene total de ceri visitantes, jsuto en pandemia, lo que podria encajar con un cierre temporal.

Asi que vamos a revisarlo.

```python
# Contamos el número de recintos arqueológicos que tienen un total de visitantes igual a cero para cada año,
# agrupando los datos del dataframe df_completo por la columna 'anio' y contando las filas donde la columna 'total' sea igual a cero

ceros_por_anio = df_completo[df_completo['total'] == 0].groupby('anio').size()
print(ceros_por_anio)
```

Con este ultimo bloque podemos ver que, los ceros tienen comportamientos distintos.
- De 1996 a 1999 tenemos de 41 a 54 registros de cero por año, loq que podrian ser muchos ceros, y luego casi desaparecen en 2000.
- Del 2000 al 2020 tenemos de 1 a 6 registros de cero por año, lo que se puede considerar como 'normal'.
- En el 2021 tenemos 60 registros de cero, un pico enorme.
- Y del 2022 al 2025 tenermos de 9 a 13 registros de ceros, mas que antes del 2020.

Una hipotesis podria ser que existe un cambio en como se realizarn los registros, por ejemplo, sitios listados pero sin conteo. Esto viendo como entre 1996-1999 tenermos de 41 a 54 registros de ceros y en 2000 baja a 1.

En 2021 tenemos 60 registros de cero, pero esto encaja con cierres por la pandemia, aunque los datos solo muestran los ceros, no el motivo.

Y en 2020 se registraron 2 ceros porque los sitios abrieron de enero a marzo.

En el describe() del bloque 18 podemos ver que es probable que la media nos engañe. Ya que la media es casi 10 veces mas que la media. Hay unos pocos sitios con millones de visitantes (el máximo es 4.18 millones en un año) que jalan el promedio hacia arriba, mientras que la mitad de los sitios recibe menos de 7.000. Así que para describir 'un sitio tipico', se deberá usar la mediana, no la media.

Con los extrangeros pasa lo mismo, la mediana 240 contra la media de 20,420. Los visitantes extrangeros se concentran en muy pocos sitios.

En los totales por año podríamos considerar tres etapas:
- Del 2012 al 2018 con un crecimiento sostenido de 10.9 a 16.7 millones.
- En 2020 cae a 4.46 millones un -72% contra el 2019.
- Y del 2022 al 2025 se estabiliza al rededor de 10 millones, con un 32% menor al 2019.

Con esto podemos tener dos matices:
- Que la cantidad de sitios que se reportan en 2019 (176) y 2025 (178) es casi igual, asi que parece que la caída no se deba a menos sitios en el conteo.
- Los datos muestran que cayó, pero no por qué. Y sin evidencia del por qué, no se debe atribuir a una causa en las conclusiones. Lo que si podemos hacer es ver qué sitios explican la diferenci.

Tambien podemos ver que nacionales y extrangeros se comportan de distinta manera, los nacionales bajaron mas entre 2019 y 2025 (10.7 a 6.3 millones) que los extranjeros (5.3 a 3.6 millones).

```python
# Creamos un nuevo dataframe llamado df_largo que contenga los datos del dataframe df_completo en formato largo,
# donde cada fila represente un mes y un tipo de visitante (nacional o extranjero)

cols_mes = cols_nac + cols_ext

df_largo = df_completo.melt(
    id_vars=['anio', 'estado', 'recinto'],
    value_vars=cols_mes,
    var_name='mes_tipo',
    value_name='visitantes'
)

df_largo.head()
```

```python
# Creamos nuevas columnas en el dataframe df_largo que representen el mes y el tipo de visitante (nacional o extranjero),
# separando la columna 'mes_tipo' en dos columnas distintas

df_largo[['mes', 'tipo']] = df_largo['mes_tipo'].str.split('_', expand=True)

df_largo['mes'] = pd.Categorical(df_largo['mes'], categories=meses, ordered=True)

print(df_largo.shape)
```

```python
# Verificamos que la suma de visitantes en el dataframe df_largo sea igual a la suma de visitantes en el dataframe df_completo

print(df_largo['visitantes'].sum() == df_completo['total'].sum())
```

```python
# Mostramos un gráfico de barras que represente el porcentaje de visitantes por mes, considerando los años 2015 a 2019

pre = df_largo[df_largo['anio'].between(2015, 2019)]

por_mes = pre.groupby('mes', observed=True)['visitantes'].sum()
por_mes_pct = por_mes / por_mes.sum() * 100

fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(x=por_mes.index, y=por_mes_pct, color='steelblue', ax=ax)
ax.set_title('¿Como se reparten las visitas durante el año? (promedio 2015-2019)')
ax.set_xlabel('Mes')
ax.set_ylabel('% de las visitas del año')
ax.axhline(100/12, linestyle='--', color='gray', label='Reparto uniforme (8.3%)')
ax.legend()
plt.xticks(rotation=45)
plt.show
```

Las visitas se concentran en marzo, diciembre y abril (~10-11% del total anual), mientas que septiembre, junio y mayo son los más bajos (~6%).
Una posible explicación son los periodos vacacionales (semana santa, verano y navidad), pero los datos no permiten confirmarlo. El patrón corresponde a 2015-2019 y está dominado por los sitios más visitados, por lo que puede no representar a un sitio tipico.

```python
# Imprimimos el porcentaje de visitantes por mes, redondeado a un decimal
# Esto para tener una idea de la distribución de visitantes a lo largo del año, considerando los años 2015 a 2019

print(por_mes_pct.round(1))
```

```python
# Aqui calculamos el porcentaje de visitantes por mes para tres periodos distintos: 2000-2009, 2015-2019 y 2022-2025
# Esto nos permite ver si hay cambios en la distribución de visitantes a lo largo del año en diferentes periodos

periodos = {'2000-2009': (2000, 2009), '2015-2019': (2015, 2019), '2022-2025': (2022, 2025)}
resultado = {}
for nombre, (a, b) in periodos.items():
    sub = df_largo[df_largo['anio'].between(a, b)]
    s = sub.groupby('mes', observed=True)['visitantes'].sum()
    resultado[nombre] = s / s.sum() * 100
pd.DataFrame(resultado).round(1)
```

El resultado del bloque anterior demuestra que el resultado de la gráfica anterior podría cambiar si se añaden los demás años al análisis. 
Ya que marzo pasa de concentrar casi el 1 de cada 5.7 visitas anuales a una de cada 10, lo que presenta un cambio enorme y es el más llamativo.
Mayo, junio y septiembre son siempre los más bajos y abril, julio y agosto se mueven un poco entre los periodos.

También se puede ver que marzo coincide con lo que pensábamos de que marzo es el mes fuerte de Teotihuacan, y en 2008 estuvo casi duplicado. Ese sitio tiene mucho peso en el total, así que podría estar inflando marzo en la década de 2000. Pero esta hipótesis podríamos comprobarla con el siguiente código.

```python
sin_teo = df_largo[~df_largo['recinto'].str.contains('Teotihuacan')]
sub = sin_teo[sin_teo['anio'].between(2000, 2009)]
s = sub.groupby('mes', observed=True)['visitantes'].sum()
print((s / s.sum() * 100).round(1))
```

```python
# Mostramos un gráfico de barras que represente el porcentaje de visitantes por mes y por tipo (nacional o extranjero), considerando los años 2015 a 2019

por_mes_tipo = (pre.groupby(['mes', 'tipo'], observed=True)['visitantes']
                    .sum().unstack())
por_mes_tipo_pct = por_mes_tipo / por_mes_tipo.sum() * 100

ax = por_mes_tipo_pct.plot(kind='bar', figsize=(10, 5))
ax.set_title('Temporada de visitantes nacionales vs extranjeros (2015-2019)')
ax.set_xlabel('Mes')
ax.set_ylabel('% de las visitas anuales del grupo')
plt.xticks(rotation=45)
plt.show()
```

En esta ultima grafica, los nacionales concentran más su año en marzo, abril y diciembre, los extranjeros lo reparten más parejo y pesan más en enero, febrero, junio y noviembre.
Esto es consistente con la idea de que los nacionales aprovechan más los periodos vacacionales, mientras que los extranjeros reparten sus visitas a lo largo del año.

```python
# Imprimimos el porcentaje de visitantes extranjeros por mes, redondeado a un decimal
# Que nos permite ver si los visitantes extranjeros tienen un patrón distinto al de los nacionales a lo largo del año, considerando los años 2015 a 2019

total_mes = por_mes_tipo.sum(axis=1)
pct_ext_por_mes = (por_mes_tipo['ext'] / total_mes * 100).round(1)
print(pct_ext_por_mes)
```
